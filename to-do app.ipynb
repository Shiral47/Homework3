# One Creative Use for an Everyday Object

This beginner-friendly notebook asks you for an everyday object and uses the OpenAI API to suggest one unusual, creative, practical use. Run each cell from top to bottom.

## 2. Enter your API key securely

The key is requested at runtime with `getpass`, so it is not shown while you type. Do not put your key in notebook code, outputs, or a GitHub commit.

```python
from getpass import getpass
from openai import OpenAI

# Ask for the key without displaying the characters as they are typed.
api_key = getpass('Enter your OpenAI API key: ')

# Create an API client for this notebook session.
client = OpenAI(api_key=api_key)
```

## 3. Choose an everyday object

Type an object you have around you, such as a mug, paperclip, or umbrella.

```python
# input() pauses and waits for you to type an object.
everyday_object = input('Enter an everyday object: ').strip()

if not everyday_object:
    raise ValueError('Please enter an everyday object before continuing.')
```

## 4. Ask for one creative idea

The Responses API will return one unusual, creative, and practical way to use your object.

```python
# Request exactly one useful idea for the object you entered.
response = client.responses.create(
    model='gpt-6-luna',
    input=(
        f'Give exactly ONE unusual, creative, and practical use for a(n) {everyday_object}. ' 
        'Explain the idea briefly and clearly.'
    ),
)

# Print the model's text answer.
print(response.output_text)
```
